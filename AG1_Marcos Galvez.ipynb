## Torres de Hanoi con Divide y vencerás

```python
def Torres_Hanoi(N, desde, hasta):
  if N ==1 : # Caso base - mover una ficha entre columnas
    print("Lleva la ficha desde " ,desde , " hasta ", hasta )

  else: # Si hay más de una ficha, recursión hasta caso base
    Torres_Hanoi(N-1, desde, 6-desde-hasta )
    print("Lleva la ficha desde " ,desde , " hasta ", hasta )
    Torres_Hanoi(N-1, 6-desde-hasta  , hasta )


Torres_Hanoi(3, 1 , 3)
```

```python
#Sucesión_de_Fibonacci
#https://es.wikipedia.org/wiki/Sucesi%C3%B3n_de_Fibonacci
#Calculo del termino n-simo de la suscesión de Fibonacci
def Fibonacci(N:int):
  if N < 2:
    return 1
  else:
    return Fibonacci(N-1)+Fibonacci(N-2)

Fibonacci(5)
```

## Devolución de cambio por técnica voraz

```python
def cambio_monedas(N, SM):
  SOLUCION = [0]*len(SM)   #SOLUCION = [0,0,0,0,..]
  ValorAcumulado = 0

  print(f'Sistema: {SM}')
  print(f'Cambio: {N}')

  for i,valor in enumerate(SM):
    monedas =  (N-ValorAcumulado)//valor
    SOLUCION[i] = monedas
    ValorAcumulado = ValorAcumulado + monedas*valor

    if ValorAcumulado == N:
      return SOLUCION


print(f'Resultado: {cambio_monedas(15,[25,10,5,1])}')
```

## N-Reinas por técnica de vueta atrás


```python
def escribe(S):
  n = len(S)
  for x in range(n):
    print("")
    for i in range(n):
      if S[i] == x+1:
        print(" X " , end="")
      else:
        print(" - ", end="")


def es_prometedora(SOLUCION,etapa):
  #Si la solución tiene dos valores iguales no es valida => Dos reinas en la misma fila
  for i in range(etapa+1):
    if SOLUCION.count(SOLUCION[i]) > 1:
      return False

    #Verifica las diagonales
    for j in range(i+1, etapa +1 ):
      if abs(i-j) == abs(SOLUCION[i]-SOLUCION[j]) : return False
  return True



def reinas(N, solucion=[], etapa=0):
  if len(solucion) == 0:
      solucion=[0 for i in range(N)]

  for i in range(1, N+1):
    solucion[etapa] = i

    if es_prometedora(solucion, etapa):
      if etapa == N-1:
        print(solucion)
        escribe(solucion)
        print()
      else:
        reinas(N, solucion, etapa+1)
    else:
      None

    solucion[etapa] = 0

reinas(4)
```

## Viaje por el rio. Programación dinámica

```python
TARIFAS = [
[0,5,4,3,999,999,999],
[999,0,999,2,3,999,11],
[999,999, 0,1,999,4,10],
[999,999,999, 0,5,6,9],
[999,999, 999,999,0,999,4],
[999,999, 999,999,999,0,3],
[999,999,999,999,999,999,0]
]



################################################################
def Precios(TARIFAS):
################################################################
  #Total de Nodos
  N = len(TARIFAS[0])

  #Inicialización de la tabla de precios
  PRECIOS = [ [9999]*N for i in [9999]*N]
  RUTA = [ [""]*N for i in [""]*N]

  for i in range(0,N-1):
    RUTA[i][i] = i             #Para ir de i a i se "pasa por i"
    PRECIOS[i][i] = 0          #Para ir de i a i se se paga 0
    for j in range(i+1, N):
      MIN = TARIFAS[i][j]
      RUTA[i][j] = i

      for k in range(i, j):
        if PRECIOS[i][k] + TARIFAS[k][j] < MIN:
            MIN = min(MIN, PRECIOS[i][k] + TARIFAS[k][j] )
            RUTA[i][j] = k          #Anota que para ir de i a j hay que pasar por k
        PRECIOS[i][j] = MIN

  return PRECIOS,RUTA
################################################################


PRECIOS,RUTA = Precios(TARIFAS)
#print(PRECIOS[0][6])

print("PRECIOS")
for i in range(len(TARIFAS)):
  print(PRECIOS[i])

print("\nRUTA")
for i in range(len(TARIFAS)):
  print(RUTA[i])

#Determinar la ruta con Recursividad
def calcular_ruta(RUTA, desde, hasta):
  if desde == hasta:
    #print("Ir a :" + str(desde))
    return ""
  else:
    return str(calcular_ruta( RUTA, desde, RUTA[desde][hasta])) +  \
                ',' + \
                str(RUTA[desde][hasta] \
              )

print("\nLa ruta es:")
calcular_ruta(RUTA, 0,6)
```

## Actividad Adicional

Problema: encontrar los dos puntos más cercanos

### Una dimensión (con divide y vencerás)

```python
import random

def encontrar_cercanos_1D(data: list):

    if len(data) > 2:

        half = len(data) // 2
        first = data[:half]
        last = data[half:]

        if len(data) % 2 != 0: # Si no es par, incluimos el elemento de la mitad en ambos
            first = data[:half+1]
            last = data[half:]

        f_distance, f_points = encontrar_cercanos_1D(first)
        l_distance, l_points = encontrar_cercanos_1D(last)

        if f_distance <= l_distance:
            distance = f_distance
            sol = f_points
        elif l_distance <= f_distance:
            distance = l_distance
            sol = l_points

        return distance, sol
        
    else:
        return abs(data[0] - data[1]), (data[0], data[1])

print(encontrar_cercanos_1D(sorted([random.randrange(1,10000) for x in range(10000)])))
```

## Dos dimensiones

```python
import math

def encontrar_cercanos_2D(data: list):

    if len(data) > 2:

        half = len(data) // 2
        first = data[:half]
        last = data[half:]

        if len(data) % 2 != 0: # Si no es par, incluimos el elemento de la mitad en ambos
            first = data[:half+1]
            last = data[half:]

        f_distance, f_points = encontrar_cercanos_2D(first)
        l_distance, l_points = encontrar_cercanos_2D(last)

        if f_distance <= l_distance:
            distance = f_distance
            sol = f_points
        elif l_distance <= f_distance:
            distance = l_distance
            sol = l_points

        return distance, sol
        
    else:
        distance = math.sqrt((data[1][0] - data[0][0])**2 + (data[1][1] - data[0][1])**2) # Distancia euclídea entre los puntos
        return distance, (data[0], data[1])
    
print(encontrar_cercanos_2D([(random.randrange(1,10000), random.randrange(1,10000)) for x in range(10000)]))
```
