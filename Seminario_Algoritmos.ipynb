Problema:
>3. Combinar cifras y operaciones

Descripción del problema:(copiar enunciado)

Disponemos de las 9 cifras del 1 al 9 (excluimos el cero) y de los 4 signos básicos de las operaciones fundamentales: suma(`+`), resta(`-`), multiplicación(`*`) y división (`/`).
Debemos combinarlos alternativamente sin repetir ninguno de ellos para obtener una cantidad dada.
Un ejemplo sería para obtener el 4: 4+2-6/3*1=4

....

(*) La respuesta es obligatoria





                                        

```python
from itertools import permutations

def combinacion(goal: int):

    numbers = range(1,10) # Todos los dígitos posibles (del 1 al 9)
    ops = ['+', '-', '*', '/'] # Los operadores disponibles en el problema

    results = []

    for n in permutations(numbers, 5): # Selección de 5 digitos dentro de los 10 posibles
        for o in permutations(ops, 4): # Selección de los operadores en diferente orden
            # Construcción de la expresión con los conjuntos de cifras y operadores
            expression = ''
            expression += ''.join(f'{n[i]}{o[i]}' for i in range(4))
            expression += f'{n[4]}'

            try:
                r = eval(expression)
            except ZeroDivisionError: # Si hay alguna división entre 0, no paramos ejecución
                continue

            if r == goal:
                results.append(expression)

    return results

def max_min():

    numbers = range(1,10) # Todos los dígitos posibles (del 1 al 9)
    ops = ['+', '-', '*', '/'] # Los operadores disponibles en el problema

    max_result = 0
    min_result = 0

    for n in permutations(numbers, 5): # Selección de 5 digitos dentro de los 10 posibles
        for o in permutations(ops, 4): # Selección de los operadores en diferente orden
            # Construcción de la expresión con los conjuntos de cifras y operadores
            expression = ''
            expression += ''.join(f'{n[i]}{o[i]}' for i in range(4))
            expression += f'{n[4]}'    

            try:
                r = eval(expression)
            except ZeroDivisionError: # Si hay alguna división entre 0, no paramos ejecución
                continue

            if r != int(r): # Si el resultado no es un número entero no nos sirve
                continue

            if r > max_result:
                max_result = r
            elif r < min_result:
                min_result = r

    return int(max_result), int(min_result)

def existe_combinaciones(max_n: int, min_n: int):
    for v in range(min_n,max_n):
        if not combinacion(v):
            return False
    return True
```

```python
from itertools import permutations

def combinacion(goal: int, numbers, ops):

    #numbers = range(1,10) # Todos los dígitos posibles (del 1 al 9)
    #ops = ['+', '-', '*', '/'] # Los operadores disponibles en el problema

    results = []

    for n in permutations(numbers, len(ops) + 1): # Selección de 5 digitos dentro de los 10 posibles
        for o in permutations(ops, len(ops)): # Selección de los operadores en diferente orden
            # Construcción de la expresión con los conjuntos de cifras y operadores
            expression = ''
            expression += ''.join(f'{n[i]}{o[i]}' for i in range(len(ops)))
            expression += f'{n[len(ops)]}'

            try:
                r = eval(expression)
            except ZeroDivisionError: # Si hay alguna división entre 0, no paramos ejecución
                continue

            if r == goal:
                results.append(expression)

    return results

resultado = combinacion(4, range(1,15), ['+', '-'])
resultado
```
